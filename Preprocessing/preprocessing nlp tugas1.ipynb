# ========== SETUP AWAL ==========

```python
import  pandas   as  pd 
import  nltk
import  re 
import  string

from  nltk.tokenize  import  word_tokenize,  sent_tokenize 
from  nltk.corpus  import  stopwords
from  Sastrawi.Stemmer.StemmerFactory  import  StemmerFactory

nltk.download('punkt') 
nltk.download('stopwords') 
nltk.download('wordnet')

factory  =  StemmerFactory()
stemmer  =  factory.create_stemmer()
stop_words_id  =  set(stopwords.words('indonesian'))
stop_words_id.update(['gw',  'ngga',  'nih',  'lah',  'sih',  'gan',  'klo'])
```

# ========== STEP 1: LOAD DATA ==========

```python
data  =  {
'id':  [1,  2,  3],
'ulasan':  ['Produk  ini  sangat  bagus   banget!  Saya  sangat  puas.',
'Barangnya  jelek,  tidak  sesuai  deskripsi  sama  sekali!!!',
'Gw  suka   banget  sama  produk  ini  nih!  Recommended!']
}
df  =  pd.DataFrame(data)
```

# ========== STEP 2: LOWERCASING ==========

```python
df['text_lower']  =  df['ulasan'].str.lower()
```

# ========== STEP 3: REMOVE PUNCTUATION ==========

```python
def  remove_punctuation(text):
    text  =  re.sub(r'[!@#$%^&*()_+=\[\]{};:\'",.<>?/\\|-]',  '',  text)
    text  =  re.sub(r'\s+',  ' ', text).strip()
    return  text

df['text_clean']  =  df['text_lower'].apply(remove_punctuation)
```

# ========== STEP 4: TOKENIZATION ==========

```python
df['tokens']  =  df['text_clean'].apply(word_tokenize)
```

# ========== STEP 5: REMOVE STOPWORDS ==========

```python
def  remove_stopwords(tokens,  stop_words):
    return  [word  for  word  in  tokens  if word.lower()  not  in  stop_words]

df['tokens_no_stop']  =  df['tokens'].apply(
lambda  x:  remove_stopwords(x,  stop_words_id))
```

# ========== STEP 6: STEMMING ==========

```python
def  apply_stemming(tokens,  stemmer):
    return  [stemmer.stem(word)  for  word  in  tokens]

df['tokens_stemmed']  =  df['tokens_no_stop'].apply(
lambda  x:  apply_stemming(x,  stemmer))
```

# ========== STEP 7: HASIL FINAL ==========

```python
df['text_final']  =  df['tokens_stemmed'].apply(lambda  x:  ' '.join(x))
```

# ========== EXPORT ==========

```python
df[['id','ulasan','text_final']].to_csv('ulasan_preprocessed.csv',  index=False)
```
